```python
import torch
import torch.nn.functional as F
from torchvision import transforms, models
from PIL import Image
import os
from sklearn.metrics import classification_report
from timm import create_model
from tqdm import tqdm
import torch.nn as nn

# --------------------------
# ✅ Device Setup
# --------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"✅ Using device: {device}")

# --------------------------
# ✅ Dataset + Class Info
# --------------------------
test_dir = "C:/Users/STUDENT/Downloads/brainmri/archive/Testing_256x256"
classes = sorted(os.listdir(test_dir))  # auto-aligns with training
print("✅ Aligned classes:", classes)

# --------------------------
# ✅ DenseNet121 Setup
# --------------------------
densenet = models.densenet121(pretrained=False)
densenet.classifier = nn.Sequential(
    nn.Dropout(p=0.5),
    nn.Linear(densenet.classifier.in_features, len(classes))
)
densenet.load_state_dict(torch.load("densenet121_best_model_weights.pth"))
densenet.to(device).eval()

# --------------------------
# ✅ InceptionV4 Setup
# --------------------------
inception = create_model("inception_v4", pretrained=False, num_classes=len(classes))
inception.load_state_dict(torch.load("C:/Users/STUDENT/rash/rashdl/inceptionv4_best_model.pth"))
inception.to(device).eval()

# --------------------------
# ✅ Image Transform
# --------------------------
transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

# --------------------------
# ✅ Collect Test Images
# --------------------------
image_paths = []
true_labels = []

for class_idx, class_name in enumerate(classes):
    class_path = os.path.join(test_dir, class_name)
    for fname in os.listdir(class_path):
        if fname.lower().endswith(('.jpg', '.jpeg', '.png')):
            image_paths.append(os.path.join(class_path, fname))
            true_labels.append(class_idx)

print(f"🔎 Found {len(image_paths)} test images.")

# --------------------------
# 🔮 Controlled Ensemble Inference
# --------------------------
temperature = 2.0
class_weights = torch.tensor([0.95, 0.95, 0.95, 0.95]).to(device)  # Slightly reduce all confidence
noise_level = 0.01

all_preds = []

for path in tqdm(image_paths, desc="🔮 Predicting"):
    try:
        img = Image.open(path).convert("RGB")
        img_tensor = transform(img).unsqueeze(0).to(device)

        with torch.no_grad():
            # Get raw logits and scale + noise
            out1 = densenet(img_tensor) / temperature + torch.randn(1, len(classes)).to(device) * noise_level
            out2 = inception(img_tensor) / temperature + torch.randn(1, len(classes)).to(device) * noise_level

            # Softmax
            prob1 = F.softmax(out1, dim=1)
            prob2 = F.softmax(out2, dim=1)

            # Weighted average ensemble
            avg_prob = 0.6 * prob1 + 0.4 * prob2

            # Optional: apply confidence adjustment
            avg_prob = avg_prob * class_weights

            pred = avg_prob.argmax(dim=1).item()

        all_preds.append(pred)
    except Exception as e:
        print(f"❌ Failed on {path}: {e}")
        all_preds.append(-1)

# --------------------------
# 📊 Evaluation
# --------------------------
valid_preds = [p for p in all_preds if p != -1]
valid_labels = [t for p, t in zip(all_preds, true_labels) if p != -1]

print("\n📋 Ensemble Classification Report:")
print(classification_report(valid_labels, valid_preds, target_names=classes))
```

```python
def predict_image(image_path):
    img = Image.open(image_path).convert("RGB")
    img_tensor = transform(img).unsqueeze(0).to(device)

    with torch.no_grad():
        out1 = densenet(img_tensor) / temperature
        out2 = inception(img_tensor) / temperature

        prob1 = F.softmax(out1, dim=1)
        prob2 = F.softmax(out2, dim=1)

        avg_prob = 0.6 * prob1 + 0.4 * prob2
        avg_prob = avg_prob * class_weights

        pred_class_idx = avg_prob.argmax(dim=1).item()
        return classes[pred_class_idx], avg_prob.cpu().numpy()

# 🔍 Example usage:
img_path = "C:/Users/STUDENT/Downloads/brainmri/archive/Testing_256x256/glioma/Te-gl_0011.jpg"
label, probs = predict_image(img_path)
print("🧠 Predicted:", label)
```
